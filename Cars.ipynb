```python

from segment_anything import sam_model_registry 
from segment_anything import SamAutomaticMaskGenerator
from segment_anything import SamPredictor
import os
import cv2
import random

import torch
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.patches as patches
%matplotlib inline
import warnings
warnings.filterwarnings("ignore")
```

```python
def show_anns(anns, axes=None):
    if len(anns) == 0:
        return
    if axes:
        ax = axes
    else:
        ax = plt.gca()
        ax.set_autoscale_on(False)
    sorted_anns = sorted(anns, key=(lambda x: x['area']), reverse=True)
    polygons = []
    color = []
    for ann in sorted_anns:
        m = ann['segmentation']
        img = np.ones((m.shape[0], m.shape[1], 3))
        color_mask = np.random.random((1, 3)).tolist()[0]
        for i in range(3):
            img[:,:,i] = color_mask[i]
            ax.imshow(np.dstack((img, m*1.0)))#m*0.5
            
        img2=np.dstack((img, m*1.0)) ### mask image
        return img2 ### get mask image
```

```python
sam_checkpoint = r"C:\Users\Abdullah\OneDrive\Desktop\archive (5)/sam_vit_h_4b8939.pth"
model_type = "vit_h" #
device = "cpu" #cpu,cuda
sam = sam_model_registry[model_type](checkpoint=sam_checkpoint)
sam.to(device=device)
mask_generator1 = SamAutomaticMaskGenerator(sam)
```

```python
path0=r'C:\Users\Abdullah\OneDrive\Desktop\archive (5)\Cars Dataset\test\Audi\47.jpg'
image = cv2.imread(path0)
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
masks1 = mask_generator1.generate(image)
bgw = np.ones(image.shape)*255
```

```python
plt.imshow(image)
plt.axis('off')
plt.show()
```

```python
#slices of mask image
for i in range(len(masks1)):
    plt.figure(figsize=(2,2))
    plt.imshow(bgw)
    show_anns([masks1[i]])
    plt.title(f'mask{i}')
    plt.axis('off')  
    plt.show()   
```

```python
m=len(masks1)//4
print(len(masks1))
fig, axs = plt.subplots(m+1,4, figsize=(8,m*2))
for i, ax in enumerate(axs.flatten()):
    if i < len(masks1):
        img=show_anns([masks1[i]])
        ax.imshow(img)
        ax.set_title(f'mask{i}')
        ax.axis('off')  
plt.tight_layout()
plt.show()
```

```python
for i in range(len(masks1)):
    
    fig, ax = plt.subplots(1,3,figsize=(6,3))

    maski=masks1[i]
    maskiseg=maski['segmentation']    
    box0=maski['bbox']
    xc,yc,w,h = box0
    x0=xc
    y0=yc
    x1=xc+w
    y1=yc+h
    
    rect = patches.Rectangle( (xc,yc),w,h, linewidth=2, edgecolor='yellow', fill=False)
    
    boximage=image[y0:y1,x0:x1,:]
    boxmask=maskiseg[y0:y1,x0:x1]
    cv2.imwrite(str(i).zfill(3)+'.png', boximage)
    np.save(str(i).zfill(3)+'.npy',boxmask)
    
    #show_anns([masksi],ax[0])
    ax[0].imshow(image)
    ax[0].add_patch(rect)
    ax[1].imshow(boximage)
    ax[2].imshow(boxmask)
    ax[0].set_title(f'mask{i}') 
    plt.show()  
```

```python
fig, ax = plt.subplots(1,2,figsize=(6,3))
boximage=cv2.imread('000.png')
maski=np.load('000.npy')
ax[0].imshow(boximage)
ax[1].imshow(maski)
ax[0].set_title(f'Original') 
ax[1].set_title(f'Mask') 
plt.show()  
```

```python
negative_img0 = np.tile(maski[:,:,np.newaxis],(1,1,3)).astype(int)
negative_img = negative_img0*255
positive_img0 = np.logical_not(negative_img)
positive_img = positive_img0.astype(np.uint8)*255
```

```python
fig, ax = plt.subplots(1,2,figsize=(6,3))
ax[0].imshow(negative_img)
ax[1].imshow(positive_img)
ax[0].set_title(f'Negative_Img') 
ax[1].set_title(f'Positive_Img') 
plt.show()  
```
